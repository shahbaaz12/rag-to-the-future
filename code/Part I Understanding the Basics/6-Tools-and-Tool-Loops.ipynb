# Tools and tool-execution loops

A model can decide that it needs a tool, but it cannot run your Python function by itself. Your application receives the requested tool call, executes the allowed tool, returns the result, and asks the model for its final answer.


````python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage, SystemMessage, ToolMessage
from langchain_core.tools import tool

# Load GROQ_API_KEY from the local .env file.
load_dotenv()

groq_model = init_chat_model(
    "openai/gpt-oss-120b",
    model_provider="groq",
)
````

## Define a small, safe tool

This example uses a local dictionary so it can run without another API key. Real tools may read a database, call an API, or perform an approved action. Keep each tool narrow and validate its inputs.


````python
@tool
def get_course_module(module_number: int) -> str:
    """Return the topic for a course module number from 1 through 6."""
    modules = {
        1: "AI foundations",
        2: "environment setup and direct model calls",
        3: "LangChain and model settings",
        4: "responses and prompts",
        5: "messages, streaming, batching, and memory",
        6: "tools and tool-execution loops",
    }
    return modules.get(module_number, "That module number is not available.")

course_tools = [get_course_module]
tools_by_name = {course_tool.name: course_tool for course_tool in course_tools}
````

## Let the model request the tool

`bind_tools` tells the model which tools it may request. The model can return a normal answer or one or more structured tool calls. It does not execute the function at this point.


````python
tool_enabled_model = groq_model.bind_tools(course_tools)

messages = [
    SystemMessage("Use the course tool when the learner asks about a module number."),
    HumanMessage("What will I learn in module 5?"),
]

model_message = tool_enabled_model.invoke(messages)
print("Requested tool calls:", model_message.tool_calls)
````

## Execute requested tools and return the result

The application controls the loop. It decides which tools are available, runs only recognised requests, and sends a `ToolMessage` containing each result back to the model.


````python
messages.append(model_message)

for tool_call in model_message.tool_calls:
    requested_tool = tools_by_name.get(tool_call["name"])
    if requested_tool is None:
        # Do not execute a tool that the application did not explicitly allow.
        tool_result = "This tool is not available."
    else:
        tool_result = requested_tool.invoke(tool_call["args"])

    messages.append(
        ToolMessage(
            content=str(tool_result),
            tool_call_id=tool_call["id"],
        )
    )

final_response = tool_enabled_model.invoke(messages)
print(final_response.text)
````

## The loop in plain English

```text
user request → model requests a tool → application validates and runs it
             → application sends result → model writes final answer
```

The model returns structured information saying that a tool could help answer the question. It does not call the Python function by itself. Our application checks the request, runs only an allowed tool, and returns the result to the model.

In a production application, add permission checks, input validation, timeouts, error handling, and logging around every tool execution.
